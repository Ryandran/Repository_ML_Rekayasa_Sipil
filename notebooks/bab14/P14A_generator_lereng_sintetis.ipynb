```python
from pathlib import Path
import numpy as np, pandas as pd
SEED=14014; rng=np.random.default_rng(SEED); out=Path("praktikum/bab14");out.mkdir(parents=True,exist_ok=True)
rows=[]
for g in range(72):
 sid=f"SLP{g+1:03d}"; angle=rng.uniform(22,48); cohesion=rng.uniform(8,35); drainage=rng.uniform(.15,.95); vegetation=rng.uniform(5,90); soil=rng.uniform(.2,1.0); latent=rng.normal(0,1.2); rains=rng.gamma(2.2,22,18)
 for t in range(18):
  rain=float(rains[t]); pore=float(np.clip(.22+.0038*rain+.10*(1-drainage)+rng.normal(0,.07),0,1.4)); season=np.sin(2*np.pi*t/12)
  # Pedagogical equation only; not a validated geotechnical simulator.
  y=max(0,.08+.0028*rain+.72*pore+.006*(angle-30)+.0018*(25-cohesion)+.001*(55-vegetation)+.10*(1-drainage)+.045*season+latent+rng.normal(0,.12))
  rows.append(dict(slope_id=sid,campaign_index=t+1,slope_angle_deg_synth=angle,cohesion_kPa_synth=cohesion,drainage_score_synth=drainage,vegetation_cover_pct_synth=vegetation,rain_7d_mm_synth=rain,pore_pressure_index_synth=pore,soil_state_index_synth=soil,deformation_response_synth_mm=y,synthetic_data=True))
df=pd.DataFrame(rows);df.to_csv(out/'slope_campaign_synthetic.csv',index=False)
cols=list(df.columns);types=[];flags=[]
for c in cols:
 if c in ('slope_id','synthetic_data'):types.append('string' if c=='slope_id' else 'discrete');flags.append('meta')
 elif c=='deformation_response_synth_mm':types.append('continuous');flags.append('class')
 else:types.append('continuous');flags.append('')
with (out/'slope_campaign_orange.tab').open('w',encoding='utf-8',newline='') as f:
 f.write('\t'.join(cols)+'\n'+'\t'.join(types)+'\n'+'\t'.join(flags)+'\n');df.to_csv(f,sep='\t',index=False,header=False)
print('seed',SEED,'rows',len(df),'slopes',df.slope_id.nunique(),'campaigns',df.campaign_index.nunique());print(df.deformation_response_synth_mm.describe().round(3));print('SYNTHETIC ONLY')
```
