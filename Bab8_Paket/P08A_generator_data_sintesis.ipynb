# P08A — Generator data sintesis

Data latihan sepenuhnya sintetis, bukan bukti/kebijakan lapangan. Seed tetap.

```python
from pathlib import Path
import numpy as np, pandas as pd
rng=np.random.default_rng(20261116); n=2400
iri0=np.clip(rng.normal(4,1,n),1.2,7.2); truck=np.clip(rng.beta(2.5,7,n),.03,.75); rain=np.clip(rng.normal(1800,420,n),700,3000)
pdrai=1/(1+np.exp(-(-1.2+.42*(iri0-4)+1.2*truck+.00035*(rain-1800)))); drain=rng.binomial(1,pdrai)
pt=1/(1+np.exp(-(-1.3+.85*(iri0-4)+1.3*drain+1.1*truck+.0004*(rain-1800)))); A=rng.binomial(1,pt)
Y=np.clip(iri0+.35+.48*(iri0-4)+.72*truck+.00045*(rain-1800)+.55*drain-.80*A+rng.normal(0,.45,n),.5,12)
df=pd.DataFrame({'site_id':[f'R{i+1:04d}' for i in range(n)],'iri_baseline':iri0,'truck_share':truck,'annual_rainfall_mm':rain,'drainage_risk':drain,'maintenance':A,'iri_6_months':Y})
out=Path('Bab8_Paket/Data'); out.mkdir(parents=True,exist_ok=True); f=out/'maintenance_synthetic.csv'; df.to_csv(f,index=False)
orange=df.copy(); orange['maintenance']=orange['maintenance'].map({0:'Tidak dipelihara',1:'Dipelihara'})
fo=out/'maintenance_orange.csv'; orange.to_csv(fo,index=False)
print('seed=20261116 rows=',len(df),'treatment=',round(df.maintenance.mean(),3),'python=',f,'orange=',fo)
```
